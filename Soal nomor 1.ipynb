```python
import hashlib
```

**Brute-force plain**

```python
target = "0110b96270248f746ecca06f1ce09746"

for i in range(1000000):
    pin = f"{i:06d}"
    h = hashlib.md5(pin.encode()).hexdigest()
    if h == target:
        print("PIN ditemukan:", pin)
        break
```

**Brute-force + unknown salt (1 digit)**

```python
target = "8d4c531eb4b0f54df72aa6839abbeaec"

for salt in "0123456789":
    for i in range(1000000):
        pin = f"{i:06d}"
        if hashlib.md5((salt + pin).encode()).hexdigest() == target:
            print(f"PIN: {pin}, Salt: {salt}")
            break
```

**Brute-force + unknown salt (1 digit) + KDF (3×)**

```python
target = "b20f8b0405b88e2b0b50eb3356d34ba7"

for salt in "0123456789":
    for i in range(1000000):
        pin = f"{i:06d}"
        data = (salt + pin).encode()
        h = hashlib.md5(data).hexdigest()  # h1
        h = hashlib.md5(h.encode()).hexdigest()  # h2
        h = hashlib.md5(h.encode()).hexdigest()  # h3
        if h == target:
            print(f"PIN: {pin}, Salt: {salt}")
            break
```

**Dictionary attack (wordlist)**

```python
target = "0110b96270248f746ecca06f1ce09746"

with open("pinlist.txt", "r") as f:
    for line in f:
        pin = line.strip()
        if hashlib.md5(pin.encode()).hexdigest() == target:
            print(f"PIN ditemukan di wordlist: {pin}")
            break
```
